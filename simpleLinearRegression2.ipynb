```python
#linear regression parameters
#fit_intercept >> The best fit line will have intercept, by default it is true
#copy_X >> copy the original X_train and then build the model, it will not modify the original data
#n_jobs >>-1 means using all processors.
#positive>> When set to True, forces the coefficients to be positive.

#fit>> train the model
#predict>> test the model

#after building the model, you have attributes of the model
#coef_
#intercept
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=1)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
x_train
```

```python
x_test
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model
```

```python
model.fit(x_train,y_train)
```

```python
model.coef_
```

```python
model.intercept_
```

```python
print("The slope and coeff of this model without scaling", model.coef_)
print("The intercept of this model without scaling", model.intercept_)
```

```python
# This coef_ and intercept_ declare without scaling.
# Now we can see that coef_ and intercept_ value after doing scaling
```

```python
scaled_model = LinearRegression()
```

```python
scaled_model
```

```python
# Scaling 
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
scaler
```

```python
#scaling should be done always after train test split
#target variableshould not be scaled
#avoid scaling categorical features
x_train = scaler.fit_transform(x_train) #for train data use fit_transform
x_test = scaler.transform(x_test) #for test data use transform
```

```python
x_train
```

```python
scaled_model
```

```python
scaled_model.fit(x_train,y_train)
```

```python
scaled_model.coef_
```

```python
scaled_model.intercept_
```

```python
print("The slope and coeff of this model with scaling", scaled_model.coef_)
print("The intercept of this model with scaling", scaled_model.intercept_)
print("The slope and coeff of this model without scaling", model.coef_)
print("The intercept of this model without scaling", model.intercept_)
```

```python
#prediction on train data with scaling
#interpretation of coef>> one unit increase in weight, on average the height increase by 17.96 units
#predicted height(y) on Xtrain = 159 + 17.96 * (Xtrain)

#predicted height(y) on Xtest = 159 + 17.96 * (Xtest)
```

```python
#prediction on train data without scaling
#interpretation of coef>> one unit increase in weight, on average the height increase by 1.08 units
#predicted height(y) on Xtrain = 77 + 1.08 * (Xtrain)

#predicted height(y) on Xtest = 77 + 1.08 * (Xtest)
```
