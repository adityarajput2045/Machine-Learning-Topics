```python
# Imputation
```

```python
titanic_datasets
```

```python
titanic_datasets.isnull().sum()
# Age is missing at random
```

```python
sns.histplot(titanic_datasets.age)
```

```python
sns.histplot(titanic_datasets.age, kde=True)
```

```python
#whenever you have normal distribution data and no outlier, impute with mean
titanic_datasets['mean_imputation'] = titanic_datasets['age'].fillna(titanic_datasets['age'].mean())
```

```python
titanic_datasets['mean_imputation']
```

```python
titanic_datasets[['age','mean_imputation']]
```

```python
sns.distplot(titanic_datasets['mean_imputation'], kde = True)
```

```python

titanic_datasets['median_imputation'] = titanic_datasets['age'].fillna(titanic_datasets['age'].median())
```

```python
titanic_datasets[['age','mean_imputation','median_imputation']]
```

```python


sns.displot(titanic_datasets.median_imputation, kde=True)
```
