```python
# Missing Value Treatment
# Import numpy and pandas

import numpy as np
import pandas as pd
```

```python
# Create random datsets
data = {
    'A' : [12,15, 100, 21, 10],
    'B' : [3, np.nan, 7, 8, 9],
    'C' : [np.nan, 12, 13, 14, 15],
    'D' : [16, 17, 18, np.nan, 20]
}

df = pd.DataFrame(data)
```

```python
df
```

```python
df.isnull()
```

```python
df
```

```python
df.isnull().sum()
```

```python
# Want to drop null row
df.dropna()
```

```python
df
```

```python
# Want to drop null coloumn
df.dropna(axis=1)
```

```python
df['B'].mean()
```

```python
df['B'].fillna(df['B'].mean())
```

```python
df
```

```python
df['B'].fillna(df['B'].median())
```

```python
df['C'].fillna(0)
```

```python
df['C'].fillna("Some String")
```
